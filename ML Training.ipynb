```python
import pandas as pd
import numpy as np
df_raw = df.copy()
df_work = df_raw.copy()
```

```python
df_work = df_work.rename(columns={"Description": "ProductDescription"})
```

```python
df_work["Revenue"] = df_work["Quantity"] * df_work["UnitPrice"]
```

```python
inv = df_work["InvoiceNo"].astype(str)
df_work["TransactionType"] = np.select(
    [
        inv.str.startswith("C"),
        df_work["Quantity"] < 0,
        (df_work["Quantity"] > 0) & (df_work["UnitPrice"] > 0),
    ],
    ["Cancellation", "Return", "Positive Sale"],
    default="Other/Invalid",
)
```

```python
df_work["IsIdentifiedCustomer"] = df_work["CustomerID"].notna()
```

```python
df_work["InvoiceDate"] = pd.to_datetime(df_work["InvoiceDate"])
```

```python
for col in ["InvoiceNo", "StockCode", "Country"]:
    df_work[col] = df_work[col].astype("string")
```

```python
df_work = df_work[[
    "InvoiceNo", "InvoiceDate", "StockCode",
    "ProductDescription", "Country",
    "CustomerID", "IsIdentifiedCustomer",
    "Quantity", "UnitPrice", "Revenue", "TransactionType",
]]
```

```python
df_work = df_work.sort_values(["InvoiceDate", "InvoiceNo"])
```

```python
df_work = df_work.reset_index(drop=True)
print(df_work.index.equals(pd.RangeIndex(len(df_work)))) 
print(df_work.index[:5].tolist(), df_work.index[-1])
```

```python
def missing_report(d):
    return pd.DataFrame({
        "Column": d.columns,
        "MissingCount": d.isnull().sum().values,
        "NonMissingCount": d.notnull().sum().values,
        "MissingPct": (d.isnull().mean() * 100).round(2).values,
    })
print(missing_report(df_work))
```

```python
df_clean["ProductDescription"] = (
    df_clean["ProductDescription"].astype("string").str.strip().replace("", pd.NA)
)
code_desc = (
    df_clean.dropna(subset=["ProductDescription"])
    .groupby("StockCode")["ProductDescription"]
    .agg(lambda s: s.mode().iloc[0])
)
df_clean["ProductDescription"] = df_clean["ProductDescription"].fillna(
    df_clean["StockCode"].map(code_desc)
)
df_clean["ProductDescription"] = df_clean["ProductDescription"].fillna("UNKNOWN PRODUCT")
```
